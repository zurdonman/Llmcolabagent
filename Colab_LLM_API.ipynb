## Cómo usar la API desde tu agente/móvil

Cuando cargues un modelo en la UI, usa la **URL pública** + `/v1` como `base_url` y la **API Key** que muestra el panel.

```python
from openai import OpenAI
client = OpenAI(base_url="https://TU-URL.trycloudflare.com/v1", api_key="TU_API_KEY")
r = client.chat.completions.create(
    model="tu-modelo",
    messages=[{"role":"user","content":"Hola"}],
)
print(r.choices[0].message.content)
```
